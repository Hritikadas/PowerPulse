## Results & Analysis
Review MAE, RMSE, R² across models and interpret feature importances.


```python
import joblib, pandas as pd, matplotlib.pyplot as plt
rf = joblib.load('../src/models/randomforest.pkl')
FEATURES = ['hour','day_of_week','month','is_weekend','is_peak',
            'mcp_lag1','mcp_roll7_mean','mcp_roll7_std',
            'mcv','buy_volume','sell_volume']
importances = pd.Series(rf.feature_importances_, index=FEATURES).sort_values()
importances.plot.barh(title='Random Forest Feature Importances')
plt.tight_layout(); plt.show()
```

```python
import joblib, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv('../data/processed/master_features.csv', parse_dates=['date'])
df = df.dropna(subset=FEATURES + ['mcp'])

X = df[FEATURES]; y = df['mcp']
_, X_test, _, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

rf = joblib.load('../src/models/randomforest.pkl')
y_pred = rf.predict(X_test)

plt.figure(figsize=(14, 4))
plt.plot(y_test.values[-200:], label='Actual', alpha=0.8)
plt.plot(y_pred[-200:], label='Predicted', alpha=0.8)
plt.title('Random Forest — Predicted vs Actual MCP (last 200 rows)')
plt.ylabel('₹/kWh'); plt.legend(); plt.tight_layout(); plt.show()
```
