```python
import sys; sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv('../data/processed/master.csv', parse_dates=['date'])
print(df.shape)
df.head()
```

## 1. Daily MCP Trend


```python
daily = df.groupby(['date','exchange'])['mcp'].mean().reset_index()
px.line(daily, x='date', y='mcp', color='exchange', title='Daily Average MCP (₹/kWh)')
```

## 2. Price Distribution


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
for exchange, grp in df.groupby('exchange'):
    grp['mcp'].hist(ax=axes[0], alpha=0.6, bins=50, label=exchange)
    grp.boxplot(column='mcp', ax=axes[1])
axes[0].set_title('MCP Histogram'); axes[0].legend()
axes[1].set_title('MCP Box Plot')
plt.tight_layout(); plt.show()
```
